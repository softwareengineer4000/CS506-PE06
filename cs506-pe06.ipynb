```python
import cv2
import matplotlib.pyplot as plt
```

```python
def measureDistance(lx,ly, glx,gly,  rx,ry, grx,gry):
    # applied pythagorean theorem to calculate the distance
    dl = ((glx - lx)**2 + (gly - ly)**2) ** .5
    dr = ((grx - rx) ** 2 + (gry - ry) ** 2) ** .5

    return dl, dr
```

```python
eye_cascade = cv2.CascadeClassifier('haarcascade_eye.xml')
```

```python
img = cv2.imread('img/train1.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
```

```python
train1_img = img.copy()
train1_rect = eye_cascade.detectMultiScale(train1_img, scaleFactor=1.1, minNeighbors=5)
```

```python
print(len(train1_rect))
```

```python
file = open('img/train1.gnd', 'r')
l = file.readline()
r = file.readline()

(glx, gly) = l.split(" ")[1].split(',')
(grx, gry) = r.split(" ")[1].split(',')

print(glx, gly)
print(grx, gry)


```

```python
for(x, y, w, h) in train1_rect:
    cv2.rectangle(train1_img, (x,y), (x+w, y+h), (255, 0, 0), 1)
    

dl, dr = measureDistance(train1_rect[0].x,train1_rect[0].y, glx,gly,
                         train1_rect[1].x,train1_rect[1].y, grx,gry)

plt.imshow(train1_img)
```
